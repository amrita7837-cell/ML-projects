```python
#  Import libraries & Load Data
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv(r"C:\Users\kumar\Downloads\Online Retail.csv", encoding='latin1')
print(df.head(10))
```

```python
# Check rows and missing values
# Number of rows and columns
print("Shape:", df.shape)

# Check missing values
print("\nMissing Values:\n", df.isnull().sum())
```

```python
# Remove duplicate values
df = df.drop_duplicates()

print("Duplicates removed. New shape:", df.shape)
```

```python
#  Add Revenue Column
df['Revenue'] = df['Quantity'] * df['UnitPrice']

print(df[['Quantity', 'UnitPrice', 'Revenue']].head())
```

```python
# Top 10 selling products (based on Quantity)
top_products = df.groupby('Description')['Quantity'].sum().sort_values(ascending=False).head(10)

print("\nTop 10 Selling Products:\n", top_products)
```

```python
# Total Revenue of Store
total_revenue = df['Revenue'].sum()

print("\nTotal Revenue:", total_revenue)
```

```python
# Total number of unique customers
unique_customers = df['CustomerID'].nunique()

print("\nTotal Unique Customers:", unique_customers)
```

```python
# Average Order Value (AOV)
# Revenue per order (Invoice)
order_revenue = df.groupby('InvoiceNo')['Revenue'].sum()

avg_order_value = order_revenue.mean()

print("\nAverage Order Value:", avg_order_value)
```

```python
# Country with highest number of orders
orders_by_country = df.groupby('Country')['InvoiceNo'].nunique()

top_country = orders_by_country.idxmax()

print("\nCountry with highest orders:", top_country)
```

```python
# Products bought together
from itertools import combinations
from collections import Counter

# Group products by Invoice
basket = df.groupby('InvoiceNo')['Description'].apply(list)

pair_counts = Counter()

for items in basket:
    pair_counts.update(combinations(items, 2))

# Top 10 combinations
top_pairs = pair_counts.most_common(10)

print("\nTop Products Bought Together:\n", top_pairs)
```

```python
# Customers who purchased more than 10 items
customer_items = df.groupby('CustomerID')['Quantity'].sum()

customers_10 = customer_items[customer_items > 10]

print("\nCustomers who purchased more than 10 items:\n", customers_10)
```

```python
# Peak shopping hours
# Convert InvoiceDate to datetime
df['InvoiceDate'] = pd.to_datetime(df['InvoiceDate'])

# Extract hour
df['Hour'] = df['InvoiceDate'].dt.hour

# Count orders per hour
peak_hours = df.groupby('Hour')['InvoiceNo'].nunique().sort_values(ascending=False)

print("\nPeak Shopping Hours:\n", peak_hours.head())
```
